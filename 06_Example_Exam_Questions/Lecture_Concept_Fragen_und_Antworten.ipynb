## Generation, Training und Sampling

a) Was lernt ein Modell im Pretraining, im Instruction Tuning und im Preference Tuning?

b) Warum kann derselbe Prompt bei zwei Aufrufen verschiedene Antworten ergeben?

c) Was bewirken Temperature und Top-p?

d) Nennen Sie eine Massnahme, um zwei Prompts trotzdem verlässlich zu vergleichen.

<details><summary style="color: #B8860B;">Antwortvorschlag</summary>

a) **Pretraining:** das nächste Token vorhersagen; Ergebnis ist ein Base Model, das Texte fortsetzt. **Instruction Tuning:** an Beispielen lernen, Anweisungen zu befolgen. **Preference Tuning:** Antworten bevorzugen, die Menschen besser bewertet haben. Im Gespräch ändern sich die Gewichte nicht.

b) Das nächste Token wird zufällig aus einer Wahrscheinlichkeitsverteilung gezogen (**Sampling**).

c) **Temperature:** Unter 1 wird die Verteilung stärker auf wahrscheinliche Tokens konzentriert; über 1 wird sie flacher. Erst bei sehr kleiner Temperature nähert sich die Auswahl dem jeweils wahrscheinlichsten Token. **Top-p:** Gezogen wird nur aus den wahrscheinlichsten Tokens, die zusammen die Wahrscheinlichkeit p erreichen.

d) Modellversion und Einstellungen festhalten und mehrere Durchläufe auf denselben Daten vergleichen.

</details>

## Wissen und Halluzination

Ein Review behauptet, die Fleischbällchen seien italienisch. Eine RAG-Antwort gibt diese Behauptung wieder und zitiert den Review korrekt. Ist die Aussage der Antwort damit wahr? Begründen Sie.

<details><summary style="color: #B8860B;">Antwortvorschlag</summary>

Nein. Die Antwort ist **faithful**: Sie gibt die Quelle richtig wieder. Die Quelle selbst kann aber falsch sein. Reviews sind persönliche Berichte und keine verlässliche Quelle für Fakten; dafür braucht es z.B. die offizielle Produktinformation.

</details>

## Tokenbudget und Reasoning

a) Warum ist die Anzahl Wörter kein verlässliches Mass für die Kosten eines Aufrufs?

b) Welche Tokens zählen zum Kontextfenster, und welche werden verrechnet?

c) Lohnt sich ein hoher Reasoning-Aufwand immer?

<details><summary style="color: #B8860B;">Antwortvorschlag</summary>

a) Verrechnet wird die Anzahl Tokens, nicht die Anzahl Wörter. Die Anzahl Tokens hängt vom Tokenizer, von der Sprache und von der Schreibweise ab.

b) Zum Kontextfenster zählen alle Input-Tokens (Instruktionen, History, Daten, Frage) und alle Output-Tokens, einschliesslich der Reasoning Tokens. Verrechnet werden Input- und Output-Tokens zu unterschiedlichen Preisen; die Reasoning Tokens sind in der Anzahl Output-Tokens bereits enthalten.

c) Nein. Höherer Reasoning-Aufwand kann mehr Output-Tokens, Kosten und Antwortzeit verursachen. Ob die Qualität steigt, muss gemessen werden.

</details>

## Gold Labels und Cohen’s Kappa

Zwei Personen labeln unabhängig voneinander dieselben 12 Reviews und vergeben bei 9 Reviews dasselbe Label. Person A vergibt die Labels `positive`, `negative` und `mixed` mit den Anteilen 0.5, 0.25 und 0.25, Person B mit den Anteilen 0.25, 0.5 und 0.25.

a) Berechnen und interpretieren Sie Cohen’s Kappa.

b) Warum genügt es nicht, einen Prompt nur gegen Silver Labels zu messen?

<details><summary style="color: #B8860B;">Antwortvorschlag</summary>

a) Beobachtete Übereinstimmung: 9 / 12 = 0.75. Zufällig erwartet: 0.5 × 0.25 + 0.25 × 0.5 + 0.25 × 0.25 = 0.3125. Kappa = (0.75 − 0.3125) / (1 − 0.3125) = 0.64: mehr Übereinstimmung als durch Zufall (1 = vollständig, 0 = Zufall).

b) Silver Labels stammen von einem Modell und können systematische Fehler enthalten. Gemeinsame Fehler von Referenz und geprüftem Modell bleiben unentdeckt.

</details>

## Entwicklung, Test und Evidenz

a) Welche Reviews dürfen als Few-Shot-Beispiele im Prompt stehen?

b) Wie prüfen Sie, ob ein neuer Prompt besser ist als der bisherige?

c) Ein Prompt liefert je Aspekt ein Label mit einem Belegzitat aus dem Review. Wie prüfen Sie die Zitate?

<details><summary style="color: #B8860B;">Antwortvorschlag</summary>

a) Nur Reviews, die weder im Development Set noch im Test Set vorkommen.

b) Den Prompt auf dem Development Set verbessern und am Schluss die festgelegten Prompts einmal auf demselben Test Set mit denselben Kennzahlen vergleichen. Danach nicht anhand der Testresultate weiter optimieren.

c) Mit Code prüfen, ob das Zitat wörtlich im Review vorkommt. Ob das Zitat das Label inhaltlich stützt, muss eine Person beurteilen.

</details>

## RAG und Judge

a) Für eine Frage wurden 10 Reviews als relevant gelabelt; 7 davon stehen unter den Top 10 der Suche. Berechnen und interpretieren Sie Recall@10.

b) Alle in einer Antwort zitierten Review-IDs gehören zu den abgerufenen Reviews. Warum genügt diese Prüfung nicht?

c) Ein zweites Modell bewertet Zusammenfassungen (LLM-as-a-Judge). Welche Verzerrungen kennen Sie, und was tun Sie dagegen?

<details><summary style="color: #B8860B;">Antwortvorschlag</summary>

a) Recall@10 = 7 / 10 = 0.7: 70 % der relevanten Reviews stehen unter den Top 10.

b) Die Prüfung zeigt nur, dass die zitierten Reviews abgerufen wurden, nicht, dass sie die Aussage stützen. Dazu muss jede Aussage mit dem zitierten Text verglichen werden.

c) Ein Judge kann Antworten nach Position, Länge oder eigenem Stil bevorzugen. Dagegen: Reihenfolge tauschen, feste Bewertungsrubrik verwenden, eine Stichprobe mit menschlichen Urteilen vergleichen.

</details>

## Function Calling und Agents

Ein Agent soll Beschwerden aus Reviews lesen und dem Filialleiter eine E-Mail senden.

a) Was macht dabei das Modell, was macht Ihre Anwendung?

b) Mit welchen Regeln begrenzen Sie das Risiko?

c) Warum ist eine Prompt Injection hier gefährlicher als bei der Klassifikation von Reviews?

<details><summary style="color: #B8860B;">Antwortvorschlag</summary>

a) Das Modell schlägt nur vor, welches Tool mit welchen Argumenten aufgerufen werden soll. Die Anwendung prüft den Vorschlag und führt ihn aus.

b) Nur die nötigen Tools und Rechte vergeben; Aktionen mit Aussenwirkung (z.B. Senden) bestätigt eine Person; Anzahl Schritte und Kosten begrenzen; Aufrufe protokollieren.

c) In unserem Klassifikationsbeispiel ist das direkte Ergebnis ein falsches Label; nachgelagerte Entscheidungen können dadurch ebenfalls Schaden nehmen. Ein Agent mit Tools kann zusätzlich Aktionen auslösen, z.B. Daten an Dritte senden.

</details>

## Deployment und Baselines

Eine Million Reviews pro Monat sollen nach ihrer Stimmung klassifiziert werden. Nennen Sie Kriterien für drei Entscheide:

a) Language Model oder einfacheres Verfahren?

b) API-Modell oder Open-Weight-Modell im eigenen Betrieb?

c) Framework oder direkter Code mit dem SDK?

<details><summary style="color: #B8860B;">Antwortvorschlag</summary>

a) Zuerst eine einfache Baseline testen (z.B. TF-IDF mit einem klassischen Modell) und nach Qualität, Kosten, Geschwindigkeit und Erklärbarkeit vergleichen.

b) **API:** einfacher Betrieb, aber die Daten gehen an den Anbieter, der Preis und Modellversionen bestimmt. **Open Weights im eigenen Betrieb:** Die Daten können im Unternehmen bleiben, dafür braucht es eigene Hardware und eigenen Betrieb.

c) Das **SDK** genügt für einfache Abläufe. Ein **Framework** kann bei komplexen Abläufen mit Zustand, Tools und Agents helfen; Wartungsaufwand und zusätzliche Abhängigkeiten sind mit abzuwägen. Einfache RAG-Pipelines sind auch mit direktem Code gut umsetzbar.

</details>

## Anonymisierung und drei Regelwerke

a) Was muss geschehen, bevor echte Reviews an eine externe API gesendet werden?

b) Worin unterscheiden sich das Schweizer Datenschutzgesetz (DSG), der EU AI Act und die Nutzungsbedingungen von Datensatz und API?

<details><summary style="color: #B8860B;">Antwortvorschlag</summary>

a) Angaben, die eine Person identifizieren, lokal entfernen und das Ergebnis stichprobenweise kontrollieren. Reicht das nicht, synthetische Daten verwenden oder lokal verarbeiten.

b) **DSG:** Bearbeitung von Personendaten. **EU AI Act:** KI-Systeme nach Einsatzzweck und Risiko. **Nutzungsbedingungen:** vertragliche Pflichten gegenüber dem Anbieter von Datensatz bzw. API. Welche Pflichten gelten, hängt von Daten, Einsatz und Geltungsbereich ab; mehrere Regelwerke können gleichzeitig relevant sein.

</details>

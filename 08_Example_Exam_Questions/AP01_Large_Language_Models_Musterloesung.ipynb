### C2 (4 Punkte)

Ein Modell extrahiert aus Reviews Aspekte (z.B. Personal, Wartezeit) und liefert zu jedem Aspekt ein Belegzitat (`evidence`), das wörtlich aus dem Review stammen soll. `quotes_C2` enthält vier solche Zitate zu zwei Reviews (simulierte Modellausgabe).

Die Teilaufgaben sind unabhängig voneinander lösbar.

**a) Code ergänzen (2 Punkte):** Prüfen Sie für jedes Zitat, ob es wörtlich im Text des Reviews vorkommt.

**b)** Ein Zitat kommt wörtlich im Review vor. Ist damit belegt, dass der extrahierte Aspekt und seine Stimmung stimmen? Begründen Sie kurz. (1 Punkt)

**c)** Das JSON Schema für die Extraktion erlaubt beim Feld `wait_minutes` (Wartezeit in Minuten) eine Zahl oder `null`. Warum ist `null` als zulässiger Wert wichtig? (1 Punkt)

**Musterantwort / Begründung:**

a) `row['evidence'] in row['text']` ist `True`, wenn das Zitat wörtlich im Text vorkommt. Zwei der vier Zitate werden gefunden (Anteil 0.5). Die beiden anderen weichen vom Text ab: andere Gross-/Kleinschreibung bzw. ein ausgelassenes Wort.

b) Nein. Die Prüfung zeigt nur, dass die Wörter im Review stehen. Ob das Zitat den Aspekt und die Stimmung inhaltlich stützt und ob Aspekte fehlen, muss eine Person am Text beurteilen.

c) Viele Reviews nennen keine Wartezeit. Ohne `null` würde das Schema eine Zahl erzwingen, und das Modell müsste einen Wert erfinden.

````python
# Vorgegeben: Belegzitate eines Modells zu zwei Reviews.
quotes_C2 = pd.DataFrame(
    {
        "review_id": ["R001", "R001", "R002", "R002"],
        "evidence": [
            "Came here on a Tuesday afternoon",
            "came here on a tuesday afternoon",
            "I have been shopping here for years",
            "I been shopping here for years",
        ],
    }
)

# Vorgegeben: zu jedem Zitat den Text des Reviews ergänzen.
df_C2 = quotes_C2.merge(df[["review_id", "text"]], on="review_id")

# Für jedes Zitat prüfen, ob es wörtlich im Text vorkommt.
found_C2 = []
for _, row in df_C2.iterrows():
    found = row["evidence"] in row["text"]
    found_C2.append(found)

# Vorgegeben: Ergebnis ausgeben.
df_C2["found"] = found_C2
print(df_C2[["review_id", "evidence", "found"]].to_string(index=False))
print("Anteil gefundener Zitate:", df_C2["found"].mean())
````
